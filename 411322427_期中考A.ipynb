````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 第 2 格：建立 SQLite 資料庫、資料表，並加入三場活動的初始資料

DB_NAME = "activity.db"

# 建立資料庫連線
conn = sqlite3.connect(DB_NAME)
cursor = conn.cursor()

try:
    # 每次執行都先刪除舊資料表
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS activities")

    # 建立活動資料表
    cursor.execute("""
        CREATE TABLE activities (
            activity_id INTEGER PRIMARY KEY,
            activity_name TEXT NOT NULL UNIQUE,
            activity_date TEXT NOT NULL,
            max_people INTEGER NOT NULL
        )
    """)

    # 建立報名資料表
    cursor.execute("""
        CREATE TABLE registrations (
            registration_id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            activity_id INTEGER NOT NULL,
            FOREIGN KEY (activity_id) REFERENCES activities(activity_id),
            UNIQUE (student_name, activity_id)
        )
    """)

    # 加入三場活動的初始資料
    activities_data = [
        (1, "迎新晚會", "2026-11-20", 6),
        (2, "程式工作坊", "2026-11-25", 4),
        (3, "籃球友誼賽", "2026-11-28", 3)
    ]

    cursor.executemany("""
        INSERT INTO activities
        (activity_id, activity_name, activity_date, max_people)
        VALUES (?, ?, ?, ?)
    """, activities_data)

    conn.commit()

    print("資料庫建立成功！")
    print("已加入三場活動，報名資料表目前為空。")

except sqlite3.Error as e:
    print("建立資料庫時發生錯誤：", e)

finally:
    conn.close()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 第 3 格：讀取 SQLite 資料表，並使用 Pandas 顯示活動與報名資料

conn = sqlite3.connect(DB_NAME)

try:
    # 顯示活動資料表
    activities_df = pd.read_sql_query(
        "SELECT * FROM activities",
        conn
    )

    # 顯示報名資料表
    registrations_df = pd.read_sql_query(
        "SELECT * FROM registrations",
        conn
    )

    print("【活動資料表 activities】")
    display(activities_df)

    print("【報名資料表 registrations】")
    display(registrations_df)

except sqlite3.Error as e:
    print("讀取資料表時發生錯誤：", e)

finally:
    conn.close()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 第 4 格：建立 AI 自然語言解讀函式，並加入結果暫存與錯誤處理

# 儲存 AI 已經解讀過的句子
ai_cache = {}

# AI 可以使用的動作
ALLOWED_ACTIONS = [
    "報名",
    "取消",
    "查詢",
    "統計",
    "無法判斷"
]

# AI 可以使用的活動名稱
ALLOWED_ACTIVITIES = [
    "迎新晚會",
    "程式工作坊",
    "籃球友誼賽",
    "全部",
    "未知"
]


def interpret_sentence(user_sentence):
    """
    將使用者的自然語言轉換成：
    動作,活動,人數
    """

    # 如果這句話之前已經解讀過，就直接使用暫存結果
    if user_sentence in ai_cache:
        return ai_cache[user_sentence]

    prompt = f"""
你是一個系學會活動報名系統的語言理解助手。

請判斷使用者的句子，並且只能輸出一行：
動作,活動,人數

【允許的動作】
報名、取消、查詢、統計、無法判斷

【允許的活動】
迎新晚會、程式工作坊、籃球友誼賽、全部、未知

【判斷規則】
1. 報名或參加活動，動作輸出「報名」，人數輸出 1。
2. 取消活動報名，動作輸出「取消」，人數輸出 1。
3. 查詢單一活動剩餘名額，動作輸出「查詢」，人數輸出 0。
4. 查詢所有活動統計，動作輸出「統計」，活動輸出「全部」，人數輸出 0。
5. 無法判斷時，輸出「無法判斷,未知,0」。
6. 不可以猜測活動名稱。
7. 不可以編造報名人數或剩餘名額。
8. 只輸出結果，不要解釋。

【使用者說的話】
{user_sentence}
"""

    # 最多呼叫 AI 三次
    max_attempts = 3
    raw_reply = ""

    for attempt in range(max_attempts):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )

            raw_reply = response.choices[0].message.content.strip()

            # 每次呼叫 AI 後等待 2 秒
            time.sleep(2)

            break

        except Exception as e:
            error_message = str(e)

            print("AI 呼叫發生錯誤：", error_message)

            # 如果是 429 使用量限制，等待 30 秒後重試
            if "429" in error_message and attempt < max_attempts - 1:
                print("使用量超過限制，等待 30 秒後重試...")
                time.sleep(30)
                continue

            # 其他錯誤或最後一次重試失敗
            raw_reply = "無法判斷,未知,0"
            time.sleep(2)
            break

    # -----------------------------
    # 不直接相信 AI 的回覆
    # -----------------------------

    # 預設值
    action = "無法判斷"
    activity = "未知"
    people = 1

    # 從 AI 回覆中找出允許的動作
    for allowed_action in ALLOWED_ACTIONS:
        if allowed_action in raw_reply:
            action = allowed_action
            break

    # 從 AI 回覆中找出允許的活動
    for allowed_activity in ALLOWED_ACTIVITIES:
        if allowed_activity in raw_reply:
            activity = allowed_activity
            break

    # 如果沒有找到允許的活動，設定為未知
    if activity not in ALLOWED_ACTIVITIES:
        activity = "未知"

    # 使用正規表示式尋找數字
    numbers = re.findall(r"\d+", raw_reply)

    if numbers:
        people = int(numbers[0])
    else:
        people = 1

    # 由程式檢查數量是否符合規格
    allowed_people = {
        "報名": 1,
        "取消": 1,
        "查詢": 0,
        "統計": 0,
        "無法判斷": 0
    }

    # 不符合動作規定的數量，就改成正確數量
    if people != allowed_people[action]:
        people = allowed_people[action]

    # 如果活動是未知，不能執行任何資料庫操作
    if activity == "未知":
        action = "無法判斷"
        people = 0

    # 儲存原始回覆與系統解讀
    result = {
        "raw": raw_reply,
        "action": action,
        "activity": activity,
        "people": people
    }

    # 暫存結果，避免相同句子重複呼叫 AI
    ai_cache[user_sentence] = result

    return result


print("AI 解讀函式建立完成！")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 第 5 格：測試 AI 是否能正確解讀 9 句自然語言

# 測試資料格式：
# (使用者說的話, 預期動作, 預期活動, 預期數量)

ai_test_cases = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),
    ("我要參加程式工作坊", "報名", "程式工作坊", 1),
    ("幫我報名籃球友誼賽", "報名", "籃球友誼賽", 1),
    ("取消迎新晚會", "取消", "迎新晚會", 1),
    ("取消程式工作坊的報名", "取消", "程式工作坊", 1),
    ("迎新晚會還有幾個名額", "查詢", "迎新晚會", 0),
    ("查詢籃球友誼賽剩餘名額", "查詢", "籃球友誼賽", 0),
    ("幫我查看所有活動的報名統計", "統計", "全部", 0),
    ("今天天氣真好", "無法判斷", "未知", 0)
]

passed = 0

for i, test_case in enumerate(ai_test_cases, start=1):

    # 取得測試資料
    sentence, expected_action, expected_activity, expected_people = test_case

    # 呼叫 AI 解讀函式
    result = interpret_sentence(sentence)

    # 取得系統解讀結果
    actual_action = result["action"]
    actual_activity = result["activity"]
    actual_people = result["people"]

    # 判斷是否符合預期
    action_ok = (
        expected_action == "*" or actual_action == expected_action
    )

    activity_ok = (
        expected_activity == "*" or actual_activity == expected_activity
    )

    people_ok = (
        expected_people == "*" or actual_people == expected_people
    )

    test_passed = action_ok and activity_ok and people_ok

    if test_passed:
        passed += 1

    print("=" * 50)
    print("測試編號：", i)
    print("使用者說的話：", sentence)
    print("AI 原始回覆：", result["raw"])

    print("系統解讀：")
    print("動作：", actual_action)
    print("活動：", actual_activity)
    print("數量：", actual_people)

    print("預期結果：", expected_action, expected_activity, expected_people)
    print("測試結果：", "通過" if test_passed else "未通過")

print("=" * 50)
print("AI 解讀測試完成！")
print("通過：", passed, "句")
print("總共：", len(ai_test_cases), "句")
print("通過率：", round(passed / len(ai_test_cases) * 100, 2), "%")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 第 6 格：建立報名、取消、查詢及統計功能

def get_connection():
    """建立資料庫連線，並啟用外鍵限制。"""

    conn = sqlite3.connect(DB_NAME)
    conn.execute("PRAGMA foreign_keys = ON")
    return conn


def process_request(student_name, user_sentence, staff_authorized=False):
    """
    根據 AI 解讀結果執行對應功能。

    student_name：使用者姓名
    user_sentence：使用者輸入的自然語言
    staff_authorized：是否已通過幹部身分驗證
    """

    # 先讓 AI 解讀使用者的句子
    result = interpret_sentence(user_sentence)

    action = result["action"]
    activity = result["activity"]

    # 無法判斷的句子不操作資料庫
    if action == "無法判斷":
        return "無法辨識您的需求，請重新輸入。"

    # 統計功能需要幹部身分驗證
    if action == "統計":
        if not staff_authorized:
            return "您沒有查看幹部統計資料的權限。"

    # 報名與取消都必須提供姓名
    if action in ["報名", "取消"]:
        if student_name is None or student_name.strip() == "":
            return "請先輸入您的姓名，才能執行報名或取消。"

        student_name = student_name.strip()

    conn = None

    try:
        conn = get_connection()
        cursor = conn.cursor()

        # ==================================
        # 功能一：報名活動
        # ==================================
        if action == "報名":

            # 確認活動是否存在
            cursor.execute("""
                SELECT activity_id, max_people
                FROM activities
                WHERE activity_name = ?
            """, (activity,))

            activity_row = cursor.fetchone()

            if activity_row is None:
                return "目前沒有此活動。"

            activity_id = activity_row[0]
            max_people = activity_row[1]

            # 檢查是否已經報名
            cursor.execute("""
                SELECT registration_id
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
            """, (student_name, activity_id))

            existing_registration = cursor.fetchone()

            if existing_registration is not None:
                return "您已經報名過此活動。"

            # 計算目前報名人數
            cursor.execute("""
                SELECT COUNT(*)
                FROM registrations
                WHERE activity_id = ?
            """, (activity_id,))

            count_row = cursor.fetchone()

            if count_row is None:
                return "無法取得目前報名人數。"

            registered_count = count_row[0]

            # 檢查是否額滿
            if registered_count >= max_people:
                return "報名失敗，活動已額滿。"

            # 新增報名紀錄
            cursor.execute("""
                INSERT INTO registrations
                (student_name, activity_id)
                VALUES (?, ?)
            """, (student_name, activity_id))

            conn.commit()

            # 重新計算剩餘名額
            remaining = max_people - registered_count - 1

            return (
                f"報名成功！\n"
                f"活動：{activity}\n"
                f"剩餘名額：{remaining}"
            )

        # ==================================
        # 功能二：取消報名
        # ==================================
        elif action == "取消":

            # 確認活動是否存在
            cursor.execute("""
                SELECT activity_id, max_people
                FROM activities
                WHERE activity_name = ?
            """, (activity,))

            activity_row = cursor.fetchone()

            if activity_row is None:
                return "目前沒有此活動。"

            activity_id = activity_row[0]
            max_people = activity_row[1]

            # 查詢學生的報名紀錄
            cursor.execute("""
                SELECT registration_id
                FROM registrations
                WHERE student_name = ?
                AND activity_id = ?
            """, (student_name, activity_id))

            registration_row = cursor.fetchone()

            if registration_row is None:
                return "查無報名紀錄。"

            # 刪除報名紀錄
            cursor.execute("""
                DELETE FROM registrations
                WHERE registration_id = ?
            """, (registration_row[0],))

            conn.commit()

            # 計算取消後的剩餘名額
            cursor.execute("""
                SELECT COUNT(*)
                FROM registrations
                WHERE activity_id = ?
            """, (activity_id,))

            count_row = cursor.fetchone()

            if count_row is None:
                return "取消成功，但無法取得剩餘名額。"

            remaining = max_people - count_row[0]

            return (
                f"取消成功，名額已釋出！\n"
                f"活動：{activity}\n"
                f"剩餘名額：{remaining}"
            )

        # ==================================
        # 功能三：查詢單一活動剩餘名額
        # ==================================
        elif action == "查詢":

            # 確認活動是否存在
            cursor.execute("""
                SELECT activity_id, activity_name, activity_date, max_people
                FROM activities
                WHERE activity_name = ?
            """, (activity,))

            activity_row = cursor.fetchone()

            if activity_row is None:
                return "目前沒有此活動。"

            activity_id = activity_row[0]
            activity_name = activity_row[1]
            activity_date = activity_row[2]
            max_people = activity_row[3]

            # 計算目前報名人數
            cursor.execute("""
                SELECT COUNT(*)
                FROM registrations
                WHERE activity_id = ?
            """, (activity_id,))

            count_row = cursor.fetchone()

            if count_row is None:
                return "無法取得目前報名人數。"

            registered_count = count_row[0]
            remaining = max_people - registered_count

            return (
                f"活動名稱：{activity_name}\n"
                f"活動日期：{activity_date}\n"
                f"人數上限：{max_people}\n"
                f"目前報名人數：{registered_count}\n"
                f"剩餘名額：{remaining}"
            )

        # ==================================
        # 功能四：查詢所有活動統計
        # ==================================
        elif action == "統計":

            # 使用 LEFT JOIN 與 GROUP BY，
            # 即使活動目前沒有人報名，也會顯示該活動。
            cursor.execute("""
                SELECT
                    a.activity_id,
                    a.activity_name,
                    a.activity_date,
                    a.max_people,
                    COUNT(r.registration_id) AS registered_count
                FROM activities AS a
                LEFT JOIN registrations AS r
                    ON a.activity_id = r.activity_id
                GROUP BY
                    a.activity_id,
                    a.activity_name,
                    a.activity_date,
                    a.max_people
                ORDER BY a.activity_id
            """)

            rows = cursor.fetchall()

            if rows is None or len(rows) == 0:
                return "目前沒有活動資料。"

            output = []
            total_registered = 0

            for row in rows:
                activity_name = row[1]
                activity_date = row[2]
                max_people = row[3]
                registered_count = row[4]

                remaining = max_people - registered_count

                if max_people > 0:
                    rate = registered_count / max_people * 100
                else:
                    rate = 0

                total_registered += registered_count

                output.append(
                    f"活動：{activity_name}\n"
                    f"日期：{activity_date}\n"
                    f"人數上限：{max_people}\n"
                    f"報名人數：{registered_count}\n"
                    f"剩餘名額：{remaining}\n"
                    f"報名率：{rate:.2f}%"
                )

            output.append(
                f"所有活動總報名人次：{total_registered}"
            )

            return "\n\n".join(output)

        else:
            return "無法辨識您的需求，請重新輸入。"

    except sqlite3.Error as e:
        print("資料庫錯誤：", e)
        return "資料庫操作失敗，請稍後再試。"

    finally:
        # 無論成功或發生錯誤，都要關閉資料庫連線
        if conn is not None:
            conn.close()


print("報名、取消、查詢及統計功能建立完成！")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 第 7 格：建立主流程，並依照規格執行 10 個流程測試

# 測試開始前，先清空所有報名紀錄
conn = get_connection()

try:
    cursor = conn.cursor()
    cursor.execute("DELETE FROM registrations")
    conn.commit()

    print("已清空報名資料，開始流程測試！")

except sqlite3.Error as e:
    print("清空報名資料時發生錯誤：", e)

finally:
    conn.close()


# 流程測試資料
# 格式：(姓名, 使用者說的話, 是否為幹部)

flow_test_cases = [
    ("王小明", "我要報名迎新晚會", False),
    ("陳小華", "我要報名程式工作坊", False),
    ("林大同", "我要報名籃球友誼賽", False),
    ("張小美", "我要報名籃球友誼賽", False),
    ("李小華", "我要報名籃球友誼賽", False),
    ("黃小明", "我要報名籃球友誼賽", False),
    ("林大同", "取消籃球友誼賽", False),
    ("王小明", "查詢迎新晚會剩餘名額", False),
    ("陳小華", "查詢程式工作坊剩餘名額", False)
]

# 依序執行測試
for i, test_case in enumerate(flow_test_cases, start=1):

    student_name = test_case[0]
    sentence = test_case[1]
    is_staff = test_case[2]

    print("\n" + "=" * 50)
    print("流程測試：", i)
    print("使用者姓名：", student_name)
    print("使用者說的話：", sentence)

    # 執行主流程
    result = process_request(
        student_name,
        sentence,
        staff_authorized=is_staff
    )

    print("系統回應：")
    print(result)


print("\n" + "=" * 50)
print("10 個流程測試全部執行完畢！")


# 顯示測試完成後的資料庫內容
conn = get_connection()

try:
    final_df = pd.read_sql_query("""
        SELECT
            r.registration_id,
            r.student_name,
            a.activity_name
        FROM registrations AS r
        JOIN activities AS a
            ON r.activity_id = a.activity_id
        ORDER BY r.registration_id
    """, conn)

    print("\n【流程測試完成後的報名紀錄】")
    display(final_df)

except sqlite3.Error as e:
    print("讀取測試結果時發生錯誤：", e)

finally:
    conn.close()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：if IN_COLAB: demo.launch(share=True)
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）


# 第 8 格：使用 Gradio 建立系學會活動報名系統網頁

# 從環境變數讀取幹部密碼
# 請事先在執行環境設定 ADMIN_PASSWORD
# 第 8 格：建立系學會活動報名系統的 Gradio 網頁介面

# 設定幹部管理員密碼

# 第 8 格：建立活動報名管理看板
# 幹部不需要密碼，打開網頁即可查看所有活動統計

def get_dashboard_data():
    """
    從 SQLite 取得所有活動的報名統計，
    並整理成看板需要的資料。
    """

    conn = None

    try:
        conn = get_connection()

        # 使用 LEFT JOIN 與 GROUP BY 計算各活動報名人數
        dashboard_df = pd.read_sql_query("""
            SELECT
                a.activity_id,
                a.activity_name AS 活動名稱,
                a.activity_date AS 活動日期,
                a.max_people AS 人數上限,
                COUNT(r.registration_id) AS 已報名人數,
                a.max_people - COUNT(r.registration_id) AS 剩餘名額,
                ROUND(
                    COUNT(r.registration_id) * 100.0 / a.max_people,
                    2
                ) AS 報名率
            FROM activities AS a
            LEFT JOIN registrations AS r
                ON a.activity_id = r.activity_id
            GROUP BY
                a.activity_id,
                a.activity_name,
                a.activity_date,
                a.max_people
            ORDER BY a.activity_id
        """, conn)

        # 計算所有活動的總報名人次
        total_registered = int(
            dashboard_df["已報名人數"].sum()
        )

        total_capacity = int(
            dashboard_df["人數上限"].sum()
        )

        total_remaining = int(
            dashboard_df["剩餘名額"].sum()
        )

        if total_capacity > 0:
            total_rate = round(
                total_registered / total_capacity * 100, 2
            )
        else:
            total_rate = 0

        # 查詢所有報名紀錄
        registration_df = pd.read_sql_query("""
            SELECT
                r.registration_id AS 報名編號,
                r.student_name AS 學生姓名,
                a.activity_name AS 活動名稱,
                a.activity_date AS 活動日期
            FROM registrations AS r
            JOIN activities AS a
                ON r.activity_id = a.activity_id
            ORDER BY a.activity_id, r.registration_id
        """, conn)

        # 統計摘要
        summary = (
            f"活動總數：{len(dashboard_df)} 場　"
            f"總報名人次：{total_registered} 人　"
            f"總剩餘名額：{total_remaining} 人　"
            f"整體報名率：{total_rate}%"
        )

        return dashboard_df, registration_df, summary

    except Exception as e:
        print("看板資料讀取錯誤：", e)

        return (
            pd.DataFrame(),
            pd.DataFrame(),
            "資料讀取失敗，請稍後重新整理。"
        )

    finally:
        if conn is not None:
            conn.close()


def refresh_dashboard():
    """
    更新看板，重新從資料庫讀取最新資料。
    """

    dashboard_df, registration_df, summary = get_dashboard_data()

    return dashboard_df, registration_df, summary


def web_interface(student_name, user_sentence):
    """
    一般同學使用的報名功能。
    """

    if user_sentence is None or user_sentence.strip() == "":
        return "請輸入您的需求。"

    user_sentence = user_sentence.strip()

    # AI 解讀使用者輸入
    result = interpret_sentence(user_sentence)

    print("AI 原始回覆：", result["raw"])
    print(
        "系統解讀：",
        result["action"],
        result["activity"],
        result["people"]
    )

    # 統計功能已經移至公開看板
    if result["action"] == "統計":
        return "所有活動統計請直接查看下方的幹部管理看板。"

    # 執行報名、取消及查詢
    response = process_request(
        student_name,
        user_sentence,
        staff_authorized=False
    )

    return response


# ==================================
# 建立 Gradio 網頁
# ==================================

with gr.Blocks(title="系學會活動報名管理系統") as demo:

    gr.Markdown("""
    # 系學會活動報名管理系統

    歡迎使用活動報名系統！

    同學可以直接輸入自然語言進行報名、取消及查詢。

    幹部可以直接在下方看板查看所有活動的即時統計，
    不需要密碼登入。
    """)

    # ==================================
    # 一般同學操作區
    # ==================================

    gr.Markdown("## 一、活動報名功能")

    student_name_input = gr.Textbox(
        label="學生姓名",
        placeholder="例如：王小明"
    )

    user_sentence_input = gr.Textbox(
        label="請輸入您的需求",
        placeholder="例如：我要報名迎新晚會"
    )

    submit_button = gr.Button(
        "送出需求",
        variant="primary"
    )

    response_output = gr.Textbox(
        label="系統回應",
        lines=4
    )

    submit_button.click(
        fn=web_interface,
        inputs=[
            student_name_input,
            user_sentence_input
        ],
        outputs=response_output
    )

    # ==================================
    # 幹部活動管理看板
    # ==================================

    gr.Markdown("---")
    gr.Markdown("## 二、幹部活動管理看板")

    gr.Markdown("""
    <div style="padding:10px; background:#e8f5e9;
                border-radius:8px; color:#1b5e20;">
    看板開放直接查看，不需要管理員密碼。
    </div>
    """)

    summary_output = gr.Textbox(
        label="活動總覽",
        interactive=False
    )

    dashboard_output = gr.Dataframe(
        label="各活動報名統計",
        interactive=False
    )

    registration_output = gr.Dataframe(
        label="所有學生報名紀錄",
        interactive=False
    )

    refresh_button = gr.Button(
        "更新看板資料",
        variant="primary"
    )

    # 按下按鈕後重新讀取 SQLite
    refresh_button.click(
        fn=refresh_dashboard,
        inputs=[],
        outputs=[
            dashboard_output,
            registration_output,
            summary_output
        ]
    )

    # 網頁開啟時，先載入一次最新統計
    demo.load(
        fn=refresh_dashboard,
        inputs=[],
        outputs=[
            dashboard_output,
            registration_output,
            summary_output
        ]
    )


# ==================================
# 啟動網頁
# ==================================

if IN_COLAB:
    demo.launch(share=True)
````
